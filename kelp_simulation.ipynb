```python
# imports
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Set Variables
ncensus = 12
M = np.array([[0.543, 0.33],
              [-0.05, 0.632]])
aC, hC, aB, hB = 0.5, 0.1, 0.15, 0.4
P = np.zeros([ncensus+1, 2])

# Initialize all at same point
P[0] = [500, 100]
for k in range(ncensus):
    # randomly draw storm and heatwave severity for this month
    S_storm = np.random.uniform(0,1)
    S_heat = np.random.uniform(0,1)

    # weather term w_t, added separately from M
    w = np.array([-aC*S_storm*P[k,0] - hC*S_heat*P[k,0],
                   aB*S_storm*P[k,0] - hB*S_heat*P[k,1]])

    P[k+1] = M @ P[k] + w

#Plotting
plt.plot(P)
plt.xlabel('Time (Months)')
plt.ylabel('Kelp Biomass')
plt.legend(['Canopy', 'Below-Surface'])
plt.title('Kelp Community Biomass Model (12 months) Under Random Weather')
```
